```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import numpy as np

img = mpimg.imread('gambar1.jpg')

mosaic = [
    ["Gambar_Asli", "RGB_16Bit"],
    ["Grayscale", "Binary"],
]

ax = plt.figure(constrained_layout=True, figsize=(10,10)).subplot_mosaic(mosaic)

ax['Gambar_Asli'].imshow(img)
ax['Gambar_Asli'].set_title('Gambar Asli')

# RGB 8bit to 16bit
rgb16bit = (img.astype(np.uint32)*257).astype(np.uint16)
ax['RGB_16Bit'].imshow(rgb16bit/65535.0)
ax['RGB_16Bit'].set_title("RGB 16Bit")

# RGB to Grayscale
r = img[:,:,0].astype(np.float32)
g = img[:,:,1].astype(np.float32)
b = img[:,:,2].astype(np.float32)
grayscale = (0.299 * r + 0.587 * g + 0.114 * b).astype(np.uint8)
ax['Grayscale'].imshow(grayscale, cmap='gray', vmin=0, vmax=255)
ax['Grayscale'].set_title("Grayscale")


#Grayscale to Binary
T = 128
img_biner = np.zeros(grayscale.shape, dtype=np.uint8)
img_biner[grayscale >= T] = 1
ax['Binary'].imshow(img_biner, cmap='binary')
ax['Binary'].set_title("Biner")


np.set_printoptions(threshold=np.inf)

print(f"Ukuran Gambar: {img.shape[0]}x{img.shape[1]}")
print(f"\n=====Gambar RGB16Bit=====")
print(f"Nilai Min: {rgb16bit.min()}")
print(f"Nilai Max: {rgb16bit.max()}")
print(f"Matriks Sample 5x5:\n {rgb16bit[100:105,100:105,:]}")

print(f"\n=====Gambar Grayscale=====")
print(f"Nilai Min: {grayscale.min()}")
print(f"Nilai Max: {grayscale.max()}")
print(f"Matriks Sample 10x10:\n {grayscale[100:110,100:110]}")

print(f"\n=====Gambar Biner=====")
print(f"Nilai Min: {img_biner.min()}")
print(f"Nilai Max: {img_biner.max()}")
print(f"Matriks Sample 10x10:\n {img_biner[100:110,100:110]}")

plt.show()
```
